```python
import pyspark
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Occurence of a word in a file").getOrCreate()
```

```python
occ = spark.sparkContext.textFile("file.txt")
```

```python
word = "example"
```

```python
occ.flatMap(lambda d:d.split(' ')).filter(lambda d: d==word).count()
```
