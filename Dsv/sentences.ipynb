```python
import pyspark
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Sentences having given word").getOrCreate()
```

```python
sent = spark.sparkContext.textFile("file1.txt")
```

```python
word = 'sentence'
```

```python
sent.flatMap(lambda d:d.split('.')).filter(lambda d: word in d).collect()
```
