```python
import pyspark
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("word count file").getOrCreate()
```

```python
parRdd = spark.sparkContext.textFile("file:///home/sois/Downloads/file.txt")
```

```python
parRdd.flatMap(lambda line:line.split(" ")).map(lambda word: (word,1)).reduceByKey(lambda a,b:a+b).collect()
```
