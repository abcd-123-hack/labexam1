```python
import pyspark
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("word count").getOrCreate()
```

```python
input = ["A sample example of word count to check the number of words called word count example to or"]
```

```python
parRdd = spark.sparkContext.parallelize(input)
```

```python
parRdd.collect()
```

```python
parRdd.count()
```

```python
parRdd.flatMap (lambda d : d.split(' ')).collect()
```

```python
wRdd = parRdd.flatMap(lambda d : d.split(' '))
```

```python
wRdd.map(lambda d : (d,1)).collect()
```

```python
kvRdd = wRdd.map(lambda d : (d,1))
```

```python
kvRdd.reduceByKey(lambda a,b : a+b).collect()
```
