```python
import pyspark
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Bank").getOrCreate()
```

```python
bank = spark.sparkContext.textFile("bank.txt")
```

```python
unique_cust = bank.map(lambda x:x.split(",")).map(lambda x:x[1]).distinct().count()
print("Num of unique customers: ", unique_cust)
```

```python
unique_bank = bank.map(lambda x:x.split(",")).map(lambda x:x[0]).distinct().count()
print("Number of unique banks : ",unique_bank)
```

```python
customers_per_bank = bank.map(lambda x:x.split(",")).map(lambda x:(x[0],x[1])).distinct().map(lambda x:(x[0],1)).reduceByKey(lambda x,y:x+y)
print("number of customers per bank: ",customers_per_bank.collect())
```

```python
acc_no = "ACC001"
no_of_trans = bank.map(lambda x:x.split(",")).filter(lambda x:x[1]==acc_no).count()
print("NUmber of transaction for: ", acc_no , ":" , no_of_trans)
```

```python
no_of_cred_trans = bank.map(lambda x:x.split(",")).filter(lambda x:x[1]==acc_no).filter(lambda x:x[3]=="credit").filter(lambda x:x[2].split("-")[2] == "2020").count()
print("credit transaction: ",no_of_cred_trans)
```
