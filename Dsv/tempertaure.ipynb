```python
import pyspark
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Temperatures of cities").getOrCreate()
```

```python
temp = spark.sparkContext.textFile("temp.txt")
```

```python
max_min = temp.map(lambda x:x.split(",")).map(lambda x:float(x[1]))
print("Maximum temp : ",max_min.max())
print("Minimum temp : ", max_min.min())
```

```python
data_point = temp.map(lambda x:x.split(",")).map(lambda x:(x[2],1)).reduceByKey(lambda x,y:x+y)
print("Data point: ",data_point.collect())
```

```python
max_min_city = temp.map(lambda x:x.split(",")).filter(lambda x: x[2]=="Bangalore").map(lambda x:float(x[1]))
print("Max of bangalore: " ,max_min_city.max())
print("Min of bangalore: " ,max_min_city.min())
```
