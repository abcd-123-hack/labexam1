```python
import pyspark
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Student").getOrCreate()
```

```python
students = spark.sparkContext.textFile("students.txt")
```

```python
students.map(lambda x:x.split(",")).map(lambda x:(x[1], 1)).reduceByKey(lambda x,y:x+y).collect()
```

```python
total_students = students.count()
print("Number of students enrolled",total_students)
```

```python
gender = students.map(lambda x:x.split(",")).map(lambda x:(x[3],1)).reduceByKey(lambda x,y:x+y)
print("Number of boys and girls",gender.collect())
```

```python
institute = "MIT"
```

```python
students_inst = students.map(lambda x:x.split(",")).filter(lambda x:x[1] == institute).map(lambda x: (x[3],1)).reduceByKey(lambda x,y: x+y)
print("Number of students from selected institute: ",students_inst.collect())
```
