```python
import pyspark
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("citizen compress").getOrCreate()
```

```python
citizens = spark.sparkContext.textFile("citizen.txt")
states = spark.sparkContext.textFile("state.txt")
```

```python
rdd = states.map(lambda x: x.split(",")).map(lambda x: (x[0],x[1])).collectAsMap()
```

```python
compressed = citizens.map(lambda x: x.split(",")).map(lambda x: (x[0],
                                                                 x[1],
                                                                 x[2],
                                                                 x[3],
                                                                 rdd.get(x[4],x[4])))
```

```python
output = compressed.map(lambda x:",".join(x))
print(output.collect())
```
