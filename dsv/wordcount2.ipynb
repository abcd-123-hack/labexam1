```python
import pyspark
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Word count 2").getOrCreate()
```

```python
input = ["example one is an example of word count with combined steps, example is example"]
```

```python
parRdd = spark.sparkContext.parallelize(input)
```

```python
parRdd.collect()
```

```python
parRdd.flatMap(lambda d:d.split(' ')).map(lambda d : (d,1)).reduceByKey(lambda a,b:a+b).collect()
```
