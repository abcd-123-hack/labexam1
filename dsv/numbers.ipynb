```python
import pyspark
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Numbers mean, max , min, sum").getOrCreate()
```

```python
nums = spark.sparkContext.textFile("number.txt")
```

```python
numbers = nums.flatMap(lambda l : l.split()).map(lambda a:int(a))
```

```python
print("Maximum: ",numbers.max())
print("Minimum: ",numbers.min())
print("Sum: ",numbers.sum())
print("Mean: ",numbers.mean())
```
