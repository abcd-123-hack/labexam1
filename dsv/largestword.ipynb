```python
import pyspark
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Largest length of word").getOrCreate()
```

```python
largest = spark.sparkContext.textFile("file.txt")
```

```python
largest.flatMap(lambda d:d.split(' ')).reduce(lambda a,b:a if len(a)>len(b) else b)
```
